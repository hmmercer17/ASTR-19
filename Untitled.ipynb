```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.table import Table
```

```python
gaia = Table.read('Gaia_1kpc.fits', format = 'fits')
print(gaia.colnames)
print(gaia)
```

```python
parallax = gaia['parallax']
g = gaia['phot_g_mean_mag']
bp = gaia['phot_bp_mean_mag']
rp = gaia['phot_rp_mean_mag']

dist = 1000/parallax # distance to the stars in parcsecs

Mg = g-5*np.log10(dist/10)

plt.figure(figsize=(4,4))
plt.hexbin(bp-rp, Mg, extent = (-3,7,-5,20), gridsize=300, cmap='plasma', bins='log')
plt.xlim(-3,7)
plt.ylim(20,-5)

x_poly = [0, 1.8, -2, -2, 0]
y_poly= [7, 17, 17, 7, 7]

plt.plot(x_poly, y_poly, '-o', c='r')

plt.xlabel('bp-rp [mag]')
plt.ylabel('Mg [mag]')
```

```python
# create an algorithm to calculate Chauvenet's Criterion

from scipy.special import erf

def event_probability(x, mu=0.0, s=1.0):
    '''
    x: array-like, event magnitude 
    mu: float, mean of the Gaussian 
    s: float, standard deviation of the Gaussian
    '''
    # how many standard deviations is x away from the mean?
    N_sigma = np.abs((x-mu)/s)

    prob_upper = 0.5*(1.0 + erf(N_sigma/np.sqrt(2)))
    prob_lower = 0.5*(1.0 + erf(-1*N_sigma/np.sqrt(2)))

    prob_inside = (prob_upper - prob_lower)

    prob_outside = 1.0 - prob_inside

    return prob_outside 

def chauvenet_criterion(prior_measurements, outlier):
    '''
    inputs: 
        prior_measurements: array containing the original dataset
        outlier: float, the outlier we want to check the null hypothesis for

    returns:
        keep: boolean, True if we can reject the null hypothesis, False if we cannot reject the null hypothesis
    '''
    # 1) compute the overall probability of randomly drawing this outlier
    mean = np.mean(prior_measurements)
    std = np.std(prior_measurements)

    outlier_probability = event_probability(outlier, mean, std)

    # 2) multiply this probability by the number of samples in the dataset
        # size of sample = size of previous measurements + 1 outlier
    N = len(prior_measurements) + 1
    product = N*outlier_probability
    
    # 3) 
    if product < 0.5 and outlier_probability < 1/(2*N):
        reject = True
        return reject

    else: 
        reject = False
        return reject
```

```python
nsamples = 1000

x = np.random.normal(loc=0.0, scale = 1.0, size=nsamples)

#make an outlier
outlier_pos = 5.0 #this is a 5 sigma event (standard deviation of 5 from the mean)

answer = chauvenet_criterion(x, outlier_pos)
print(f'Can we reject the null hypothesis? {answer}')
```

```python
# example of an outlier we cant reject
nsamples2 = 10000000

x = np.random.normal(loc=0.0, scale = 1.0, size=nsamples2)

#make an outlier
outlier_pos2 = 5.0 #this is a 5 sigma event (standard deviation of 5 from the mean)

answer = chauvenet_criterion(x, outlier_pos2)
print(f'Can we reject the null hypothesis? {answer}')
```
